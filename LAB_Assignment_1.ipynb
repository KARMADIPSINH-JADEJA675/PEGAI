```python
from IPython.display import HTML

game = """
<!DOCTYPE html>
<html>
<head>
<meta charset="UTF-8">
<title>Snake Game</title>

<style>
body{
    background:#222;
    color:white;
    font-family:Arial, sans-serif;
    text-align:center;
}

h2{
    color:#00ff99;
}

canvas{
    background:#111;
    border:4px solid white;
    margin-top:10px;
}

#score{
    color:#00ffff;
    font-weight:bold;
}
</style>

</head>
<body>

<h2>🐍 Snake Game</h2>

<p>
🟦 Blue Dot = +1 Tail & +1 Score <br>
🟧 Orange Dot = -1 Tail & -1 Score
</p>

<h3>Score : <span id="score">3</span></h3>

<canvas id="game" width="400" height="400"></canvas>

<script>

const canvas = document.getElementById("game");
const ctx = canvas.getContext("2d");

const box = 20;
const rows = 20;
const cols = 20;

let snake = [
    {x:9,y:10},
    {x:8,y:10},
    {x:7,y:10}
];

let direction = "RIGHT";
let score = 3;

function randomPos(){

    while(true){

        let p={
            x:Math.floor(Math.random()*cols),
            y:Math.floor(Math.random()*rows)
        };

        let ok=true;

        for(let s of snake){
            if(s.x==p.x && s.y==p.y){
                ok=false;
                break;
            }
        }

        if(ok) return p;
    }
}

let blue = randomPos();
let orange = randomPos();

document.addEventListener("keydown",changeDirection);

function changeDirection(e){

    if(e.key=="ArrowLeft" && direction!="RIGHT")
        direction="LEFT";

    if(e.key=="ArrowUp" && direction!="DOWN")
        direction="UP";

    if(e.key=="ArrowRight" && direction!="LEFT")
        direction="RIGHT";

    if(e.key=="ArrowDown" && direction!="UP")
        direction="DOWN";
}

function collision(head,array){

    for(let i=0;i<array.length;i++){

        if(head.x==array[i].x && head.y==array[i].y)
            return true;
    }

    return false;
}

function gameOver(msg){

    clearInterval(game);

    alert(msg + "\\n\\nFinal Score : " + score);

}

function draw(){

    ctx.fillStyle="#111";
    ctx.fillRect(0,0,400,400);

    // Snake

    for(let i=0;i<snake.length;i++){

        ctx.fillStyle = (i==0) ? "#00ff00" : "#00aa00";

        ctx.fillRect(
            snake[i].x*box,
            snake[i].y*box,
            box,
            box
        );
    }

    // Blue Dot

    ctx.beginPath();
    ctx.fillStyle="blue";
    ctx.arc(
        blue.x*box+10,
        blue.y*box+10,
        7,
        0,
        Math.PI*2
    );
    ctx.fill();

    // Orange Dot

    ctx.beginPath();
    ctx.fillStyle="orange";
    ctx.arc(
        orange.x*box+10,
        orange.y*box+10,
        7,
        0,
        Math.PI*2
    );
    ctx.fill();

    let headX = snake[0].x;
    let headY = snake[0].y;

    if(direction=="LEFT") headX--;
    if(direction=="RIGHT") headX++;
    if(direction=="UP") headY--;
    if(direction=="DOWN") headY++;

    // Wall Collision

    if(headX<0 || headX>=cols || headY<0 || headY>=rows){

        gameOver("❌ Snake went outside the box!");
        return;
    }

    let newHead = {x:headX,y:headY};

    // Self Collision

    if(collision(newHead,snake)){

        gameOver("❌ Snake hit itself!");
        return;
    }

    // Blue Dot (+1)

    if(headX==blue.x && headY==blue.y){

        score++;

        document.getElementById("score").innerHTML=score;

        snake.unshift(newHead);

        blue=randomPos();

    }

    // Orange Dot (-1)

    else if(headX==orange.x && headY==orange.y){

        score--;

        document.getElementById("score").innerHTML=score;

        snake.unshift(newHead);

        // Reduce tail by one
        if(snake.length>1){
            snake.pop();
            snake.pop();
        }

        orange=randomPos();

        if(score<=0){

            gameOver("❌ Score became 0!");
            return;
        }

    }

    // Normal Movement

    else{

        snake.pop();
        snake.unshift(newHead);

    }

}

let game = setInterval(draw,150);

</script>

</body>
</html>
"""

HTML(game)
```
